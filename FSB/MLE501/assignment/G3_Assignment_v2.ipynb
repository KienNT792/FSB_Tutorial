```python
import requests as req
import pandas as pd
from datetime import datetime, timedelta
from calendar import monthrange
```

```python
def get_weather_data(start_date, end_date, api_key):
    url = f"https://api.weatherbit.io/v2.0/history/airquality"
    params = {
        'city': 'Hanoi',
        'start_date': start_date,
        'end_date': end_date,
        'tz': 'local',
        'key': api_key
    }

    response = requests.get(url, params=params, verify=False)
    if response.status_code == 200:
        data = response.json()
        records = data.get('data', [])
        df = pd.DataFrame(records)
        return df
    else:
        print(f"Error: {response.status_code}")
        return None
```

```python
def save_data_to_csv(data, file_name):
    if data is not None and 'data' in data and len(data['data']) > 0:
        records = data['data']
        print(f"Saving {len(records)} records.")  # In ra số bản ghi đang được lưu
        df = pd.DataFrame(records)
        df.to_csv(file_name, index=False, mode='a', header=not pd.io.common.file_exists(file_name))
        print(f"Data saved to {file_name}")
    else:
        print("No valid data to save.")
```

```python
def get_months_between_dates(start_date, end_date):
    months = []
    start_year, start_month = start_date.year, start_date.month
    end_year, end_month = end_date.year, end_date.month

    while start_year < end_year or (start_year == end_year and start_month <= end_month):
        # Tạo start_date và end_date cho từng tháng
        start = datetime(start_year, start_month, 1)
        _, last_day = monthrange(start_year, start_month)  # Lấy số ngày cuối tháng
        end = datetime(start_year, start_month, last_day)

        months.append((start.strftime('%Y-%m-%d'), end.strftime('%Y-%m-%d')))

        # Chuyển sang tháng tiếp theo
        if start_month == 12:
            start_month = 1
            start_year += 1
        else:
            start_month += 1

    return months
```

## 2.4 Feature Engineering

Tạo các features mới từ dữ liệu hiện có để cải thiện khả năng dự đoán PM2.5

```python
# Tạo bản sao để feature engineering
df_features = df_clean.copy()

print("=" * 80)
print("FEATURE ENGINEERING - TEMPORAL FEATURES")
print("=" * 80)

# 1. Temporal Features (Đặc trưng thời gian)
df_features['year'] = df_features['Local Time'].dt.year
df_features['month'] = df_features['Local Time'].dt.month
df_features['day'] = df_features['Local Time'].dt.day
df_features['hour'] = df_features['Local Time'].dt.hour
df_features['day_of_week'] = df_features['Local Time'].dt.dayofweek  # 0=Monday, 6=Sunday
df_features['day_of_year'] = df_features['Local Time'].dt.dayofyear
df_features['week_of_year'] = df_features['Local Time'].dt.isocalendar().week

# 2. Cyclical Features (cho hour và month)
# Hour
df_features['hour_sin'] = np.sin(2 * np.pi * df_features['hour'] / 24)
df_features['hour_cos'] = np.cos(2 * np.pi * df_features['hour'] / 24)

# Month
df_features['month_sin'] = np.sin(2 * np.pi * df_features['month'] / 12)
df_features['month_cos'] = np.cos(2 * np.pi * df_features['month'] / 12)

# Day of week
df_features['day_of_week_sin'] = np.sin(2 * np.pi * df_features['day_of_week'] / 7)
df_features['day_of_week_cos'] = np.cos(2 * np.pi * df_features['day_of_week'] / 7)

# 3. Phân loại thời gian
def get_season(month):
    if month in [12, 1, 2]:
        return 0  # Winter (Đông)
    elif month in [3, 4, 5]:
        return 1  # Spring (Xuân)
    elif month in [6, 7, 8]:
        return 2  # Summer (Hạ)
    else:
        return 3  # Fall (Thu)

df_features['season'] = df_features['month'].apply(get_season)

# 4. Phân loại giờ trong ngày
def get_time_of_day(hour):
    if 6 <= hour < 12:
        return 0  # Morning
    elif 12 <= hour < 18:
        return 1  # Afternoon
    elif 18 <= hour < 22:
        return 2  # Evening
    else:
        return 3  # Night

df_features['time_of_day'] = df_features['hour'].apply(get_time_of_day)

# 5. Cuối tuần hay không
df_features['is_weekend'] = (df_features['day_of_week'] >= 5).astype(int)

print(f"\n✓ Đã tạo {15} temporal features mới!")
print("\nTemporal features bao gồm:")
temporal_cols = ['year', 'month', 'day', 'hour', 'day_of_week', 'day_of_year', 'week_of_year',
                 'hour_sin', 'hour_cos', 'month_sin', 'month_cos', 'day_of_week_sin', 'day_of_week_cos',
                 'season', 'time_of_day', 'is_weekend']
for i, col in enumerate(temporal_cols, 1):
    print(f"  {i}. {col}")

print("\n" + "=" * 80)
```

```python
print("=" * 80)
print("FEATURE ENGINEERING - LAG FEATURES & ROLLING STATISTICS")
print("=" * 80)

# 6. Lag Features (Giá trị quá khứ của PM2.5)
lag_periods = [1, 2, 3, 6, 12, 24]  # 1h, 2h, 3h, 6h, 12h, 24h trước
for lag in lag_periods:
    df_features[f'PM25_lag_{lag}h'] = df_features['PM25'].shift(lag)

print(f"\n✓ Đã tạo {len(lag_periods)} lag features cho PM2.5")
print(f"  Lag periods: {lag_periods} giờ")

# 7. Rolling Statistics (Thống kê trượt)
window_sizes = [3, 6, 12, 24]  # 3h, 6h, 12h, 24h
for window in window_sizes:
    # Rolling mean
    df_features[f'PM25_rolling_mean_{window}h'] = df_features['PM25'].rolling(window=window, min_periods=1).mean()
    # Rolling std
    df_features[f'PM25_rolling_std_{window}h'] = df_features['PM25'].rolling(window=window, min_periods=1).std()
    # Rolling max
    df_features[f'PM25_rolling_max_{window}h'] = df_features['PM25'].rolling(window=window, min_periods=1).max()
    # Rolling min
    df_features[f'PM25_rolling_min_{window}h'] = df_features['PM25'].rolling(window=window, min_periods=1).min()

print(f"\n✓ Đã tạo {len(window_sizes) * 4} rolling statistics features cho PM2.5")
print(f"  Window sizes: {window_sizes} giờ")
print(f"  Metrics: mean, std, max, min")

# 8. Lag features cho các biến khí tượng quan trọng
weather_vars = ['Temperature', 'Relative Humidity', 'Wind Speed', 'Pressure']
for var in weather_vars:
    for lag in [1, 3, 6]:
        df_features[f'{var}_lag_{lag}h'] = df_features[var].shift(lag)

print(f"\n✓ Đã tạo {len(weather_vars) * 3} lag features cho các biến khí tượng")
print(f"  Variables: {', '.join(weather_vars)}")

print("\n" + "=" * 80)
```

```python
print("=" * 80)
print("FEATURE ENGINEERING - INTERACTION & POLLUTION FEATURES")
print("=" * 80)

# 9. Interaction Features (Tương tác giữa các biến)
# Nhiệt độ và độ ẩm
df_features['temp_humidity_interaction'] = df_features['Temperature'] * df_features['Relative Humidity']

# Gió và độ ẩm
df_features['wind_humidity_interaction'] = df_features['Wind Speed'] * df_features['Relative Humidity']

# Nhiệt độ và gió
df_features['temp_wind_interaction'] = df_features['Temperature'] * df_features['Wind Speed']

# Mây và mưa
df_features['clouds_precip_interaction'] = df_features['Clouds'] * df_features['Precipitation']

print(f"\n✓ Đã tạo 4 interaction features")

# 10. Pollution Ratio Features
# Tỷ lệ PM2.5 trên PM10
df_features['PM25_PM10_ratio'] = df_features['PM25'] / (df_features['PM10'] + 0.001)

# Tỷ lệ các chất ô nhiễm
df_features['CO_NO2_ratio'] = df_features['CO'] / (df_features['NO2'] + 0.001)
df_features['O3_NO2_ratio'] = df_features['O3'] / (df_features['NO2'] + 0.001)

print(f"✓ Đã tạo 3 pollution ratio features")

# 11. Weather Composite Features
# Heat Index (Chỉ số nóng)
df_features['heat_index'] = df_features['Temperature'] * df_features['Relative Humidity'] / 100

# Wind Chill effect
df_features['wind_chill'] = df_features['Temperature'] - (df_features['Wind Speed'] * 0.5)

# Atmospheric stability
df_features['atmospheric_stability'] = df_features['Temperature'] / (df_features['Pressure'] + 0.001)

print(f"✓ Đã tạo 3 weather composite features")

print("\n" + "=" * 80)
```

```python
print("=" * 80)
print("FEATURE ENGINEERING - POLYNOMIAL & DOMAIN FEATURES")
print("=" * 80)

# 12. Polynomial Features (bậc 2 cho một số biến quan trọng)
important_vars = ['Temperature', 'Relative Humidity', 'Wind Speed', 'Pressure']
for var in important_vars:
    df_features[f'{var}_squared'] = df_features[var] ** 2

print(f"\n✓ Đã tạo {len(important_vars)} polynomial features (bậc 2)")

# 13. Pollution Index (Chỉ số tổng hợp ô nhiễm)
# Tổng các chất ô nhiễm (chuẩn hóa)
pollutants = ['CO', 'NO2', 'O3', 'PM10', 'PM25', 'SO2']
df_features['total_pollutants'] = df_features[pollutants].sum(axis=1)
df_features['avg_pollutants'] = df_features[pollutants].mean(axis=1)

print(f"✓ Đã tạo 2 pollution index features")

# 14. Rate of Change (Tốc độ thay đổi)
df_features['PM25_change_1h'] = df_features['PM25'].diff(1)
df_features['PM25_change_3h'] = df_features['PM25'].diff(3)
df_features['temp_change_1h'] = df_features['Temperature'].diff(1)

print(f"✓ Đã tạo 3 rate of change features")

print("\n" + "=" * 80)
```

```python
# Tổng kết Features
print("=" * 80)
print("TỔNG KẾT FEATURE ENGINEERING")
print("=" * 80)

original_cols = df_clean.shape[1]
new_cols = df_features.shape[1]
added_features = new_cols - original_cols

print(f"\n📊 Số lượng features:")
print(f"  • Features ban đầu: {original_cols}")
print(f"  • Features mới tạo: {added_features}")
print(f"  • Tổng số features: {new_cols}")

print(f"\n📝 Phân loại features mới:")
print(f"  • Temporal features: 15")
print(f"  • Lag features (PM2.5): 6")
print(f"  • Rolling statistics: 16")
print(f"  • Lag features (weather): 12")
print(f"  • Interaction features: 4")
print(f"  • Pollution ratio features: 3")
print(f"  • Weather composite features: 3")
print(f"  • Polynomial features: 4")
print(f"  • Pollution index: 2")
print(f"  • Rate of change: 3")
print(f"  ────────────────────────")
print(f"  TỔNG CỘNG: {added_features} features mới")

print("\n" + "=" * 80)

# Hiển thị một vài dòng để kiểm tra
print("\nMẫu dữ liệu sau Feature Engineering:")
df_features.head(3)
```

## 2.5 Chuẩn bị Dữ liệu cho Model Training

```python
print("=" * 80)
print("XỬ LÝ MISSING VALUES VÀ CHUẨN BỊ DỮ LIỆU")
print("=" * 80)

# Kiểm tra missing values sau khi tạo features
print(f"\n1. Số lượng missing values:")
missing_count = df_features.isnull().sum().sum()
print(f"   Tổng missing values: {missing_count:,}")

# Loại bỏ các dòng có missing values (do lag và rolling features)
df_model = df_features.dropna().copy()
print(f"\n2. Sau khi loại bỏ missing values:")
print(f"   • Số dòng trước: {len(df_features):,}")
print(f"   • Số dòng sau: {len(df_model):,}")
print(f"   • Số dòng bị loại: {len(df_features) - len(df_model):,}")

# Xác định các cột không cần thiết cho model
cols_to_drop = ['Local Time', 'UTC Time', 'City', 'Country Code', 'Timezone']

# Xác định target variable
target = 'PM25'

# Xác định features
feature_cols = [col for col in df_model.columns if col not in cols_to_drop + [target]]

print(f"\n3. Cấu trúc dữ liệu cho model:")
print(f"   • Target variable: {target}")
print(f"   • Số lượng features: {len(feature_cols)}")
print(f"   • Số lượng samples: {len(df_model):,}")

# Tạo X và y
X = df_model[feature_cols].copy()
y = df_model[target].copy()

print(f"\n4. Shape của dữ liệu:")
print(f"   • X shape: {X.shape}")
print(f"   • y shape: {y.shape}")

print("\n" + "=" * 80)
```

```python
print("=" * 80)
print("CHIA DỮ LIỆU TRAIN/TEST")
print("=" * 80)

# Chia dữ liệu theo thời gian (80% train, 20% test)
# Không shuffle để giữ tính tuần tự thời gian
split_index = int(len(X) * 0.8)

X_train = X.iloc[:split_index]
X_test = X.iloc[split_index:]
y_train = y.iloc[:split_index]
y_test = y.iloc[split_index:]

print(f"\n📊 Phân chia dữ liệu:")
print(f"\n  TRAINING SET:")
print(f"    • X_train shape: {X_train.shape}")
print(f"    • y_train shape: {y_train.shape}")
print(f"    • Tỷ lệ: {X_train.shape[0]/len(X)*100:.1f}%")

print(f"\n  TEST SET:")
print(f"    • X_test shape: {X_test.shape}")
print(f"    • y_test shape: {y_test.shape}")
print(f"    • Tỷ lệ: {X_test.shape[0]/len(X)*100:.1f}%")

print(f"\n📈 Thống kê PM2.5:")
print(f"  Training set - Mean: {y_train.mean():.2f}, Std: {y_train.std():.2f}, Min: {y_train.min():.2f}, Max: {y_train.max():.2f}")
print(f"  Test set     - Mean: {y_test.mean():.2f}, Std: {y_test.std():.2f}, Min: {y_test.min():.2f}, Max: {y_test.max():.2f}")

print("\n" + "=" * 80)
```

```python
print("=" * 80)
print("DANH SÁCH CHI TIẾT CÁC FEATURES")
print("=" * 80)

# Phân loại features
feature_categories = {
    'Original Features': [col for col in feature_cols if col in df_clean.columns and col not in ['Local Time', 'UTC Time', 'City', 'Country Code', 'Timezone']],
    'Temporal Features': [col for col in feature_cols if any(x in col for x in ['year', 'month', 'day', 'hour', 'week', 'season', 'time_of_day', 'weekend', '_sin', '_cos'])],
    'Lag Features': [col for col in feature_cols if 'lag' in col],
    'Rolling Features': [col for col in feature_cols if 'rolling' in col],
    'Interaction Features': [col for col in feature_cols if 'interaction' in col],
    'Ratio Features': [col for col in feature_cols if 'ratio' in col],
    'Composite Features': [col for col in feature_cols if any(x in col for x in ['heat_index', 'wind_chill', 'atmospheric', 'total_', 'avg_'])],
    'Polynomial Features': [col for col in feature_cols if 'squared' in col],
    'Change Features': [col for col in feature_cols if 'change' in col]
}

for category, cols in feature_categories.items():
    if len(cols) > 0:
        print(f"\n📌 {category} ({len(cols)} features):")
        for i, col in enumerate(cols, 1):
            if i <= 10:  # Hiển thị tối đa 10 features đầu tiên
                print(f"   {i:2d}. {col}")
            elif i == 11:
                print(f"   ... và {len(cols) - 10} features khác")
                break

print("\n" + "=" * 80)
print(f"\n✅ DỮ LIỆU ĐÃ SẴN SÀNG CHO MODEL TRAINING!")
print(f"   • Tổng số features: {len(feature_cols)}")
print(f"   • Training samples: {len(X_train):,}")
print(f"   • Test samples: {len(X_test):,}")
print("\n" + "=" * 80)
```
